```python
from pathlib import Path
from PIL import Image

IMG_DIR = Path("images")

def load_gray(path, max_side = 512):

    im = Image.open(path).convert("L")
    im.thumbnail((max_side, max_side))

    return np.asarray(im, dtype = float) / 255.0

paths = sorted(p for p in IMG_DIR.iterdir() if p.suffix.lower() in {".jpg"})
print([p.name for p in paths])

img = load_gray(paths[0])
```

```python
import matplotlib.pyplot as plt

def gallery(paths, cols = 4, size = 3):
    rows = -(-len(paths) // cols) 
    fig, axes = plt.subplots(rows, cols, figsize = (size * cols, size * rows))

    for ax in axes.ravel():
        ax.axis("off")
        
    for ax, p in zip(axes.ravel(), paths):
        ax.imshow(load_gray(p), cmap = "gray")
        ax.set_title(p.name, fontsize = 8)
    plt.show()

gallery(paths)
```
